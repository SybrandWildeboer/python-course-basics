# Lists

A list holds things in order.

```python
devices = ["phone", "laptop", "speaker", "car"]

print(devices[0])      # phone      counting starts at 0
print(devices[2])      # speaker
print(devices[-1])     # car        the last one
print(devices[-2])     # speaker    second from the end
print(len(devices))    # 4
```

**Zero is the first.** This is why `range(5)` gave you 0 to 4 in session 2.
A list of 4 items has positions 0, 1, 2, 3, and the last position is always
`len(list) - 1`, or just `-1`.

Negative positions count backwards from the end, which saves you arithmetic.

Ask for a position that does not exist and you get a new error for your
collection. Uncomment it.

```python
# print(devices[9])
# IndexError: list index out of range
```

## Slicing: a piece of a list

```python
nums = [10, 20, 30, 40, 50, 60]

print(nums[1:4])     # [20, 30, 40]   from 1, up to but NOT 4
print(nums[:3])      # [10, 20, 30]   from the start
print(nums[3:])      # [40, 50, 60]   to the end
print(nums[-2:])     # [50, 60]       the last two
```

Same rule as `range`: the end position is not included. `nums[1:4]` gives
three items, not four.

The practical use you will want within the hour: `rows[:5]` to peek at the
first five rows of a file, instead of printing two thousand lines into your
notebook.

## Growing, changing, checking

```python
genres = ["Folk", "Jazz"]

genres.append("Rock")        # add to the end
print(genres)

genres[0] = "Indie"          # replace in place
print(genres)

print("Jazz" in genres)      # True
print("Metal" in genres)     # False
```

`in` is one word to ask "is this in there", and it reads like English inside
an `if`. You will use it constantly.

## `sorted()` versus `.sort()`

This distinction catches people, and it connects straight back to session 3.

```python
print(sorted(genres))        # a NEW sorted list
print(genres)                # the original, untouched

print(genres.sort())         # None! it sorts the list and returns nothing
print(genres)                # ...but the list itself is now sorted
```

So `x = x.sort()` throws your data away: the sort happens, and then you
overwrite `x` with `None`. It is exactly the session 3 lesson about functions
that do something and return nothing.

## Looping over a list

```python
minutes = [4.65, 2.42, 7.10, 3.05]

total = 0
for value in minutes:
    total += value

print(f"{total:.2f} minutes")
```

Same accumulate pattern as session 2: set up before the loop, update inside,
use after. It does not matter whether you are looping over `range(10)` or
over two thousand rows of real data.

When you need the position as well as the value, use `enumerate`:

```python
for i, value in enumerate(minutes):
    print(f"{i}: {value}")

for n, value in enumerate(minutes, start=1):
    print(f"row {n} is {value}")
```

`enumerate` feels like a nicety now. It becomes essential the first time an
error says "row 47" and you want to know which row that was.

# Dictionaries

A list finds things by **position**. A dictionary finds them by **name**.

```python
play = {
    "artist_name": "Fjord & Flint",
    "track_name": "Golden Garden",
    "genre": "Folk",
    "minutes_played": 4.65,
}

print(play["genre"])
print(play["minutes_played"])
```

For a row of data, names win every time. `play["genre"]` still makes sense
in six months; `play[4]` does not.

The names are called **keys**, the things they point at are **values**. Keys
are usually strings, and each one appears once.

Ask for a key that is not there and you get another error worth recognising.
It is the one you will hit most this week, usually from a mistyped column
name.

```python
# print(play["mood"])
# KeyError: 'mood'
```

## Adding, updating, and `.get()`

```python
counts = {"Folk": 438, "Jazz": 154}

counts["Rock"] = 130        # add a new key
counts["Folk"] = 439        # update an existing one
print(counts)

print("Jazz" in counts)     # True
```

```python
print(counts.get("Metal"))       # None, no crash
print(counts.get("Metal", 0))    # 0, a default you chose
```

**When to use which:**

* `[]` when a missing key means something is genuinely wrong and you want to
  hear about it
* `.get()` with a default when missing is a normal thing that happens

This line is the counting pattern in one expression, and you will write it
in today's exercise. It means "whatever is there, or zero if nothing is,
plus one".

```python
genre = "Latin"
counts[genre] = counts.get(genre, 0) + 1
print(counts)
```

## Looping over a dictionary

```python
counts = {"Electronic": 547, "Folk": 438, "Pop": 302}

for genre in counts:                    # keys, by default
    print(genre)
```

```python
for number in counts.values():          # just the values
    print(number)
```

```python
for genre, number in counts.items():    # both at once
    print(f"{genre}: {number}")
```

`.items()` is the one you will use most.

And here is the recipe for turning a dictionary of counts into a ranked
list. Copy it, use it, and do not worry about `lambda` yet: it means "sort
by the second thing in each pair".

```python
for genre, number in sorted(counts.items(),
                            key=lambda pair: pair[1],
                            reverse=True):
    print(f"{genre:12} {number:>5}")
```

`{genre:12}` pads to 12 characters and `{number:>5}` right-aligns in 5. That
is how you get columns that line up.

# The big idea: a list of dictionaries is a table

One dictionary is a **row**. One key is a **column**. A list of them is a
**table**.

```python
rows = [
    {"genre": "Folk", "minutes": 4.65},
    {"genre": "Jazz", "minutes": 7.10},
    {"genre": "Folk", "minutes": 2.42},
]

print(len(rows))              # 3 rows
print(rows[0])                # the first row
print(rows[0]["genre"])       # row 0, column "genre"
```

Which is the same information, in the same shape, as this:

| genre | minutes |
|---|---|
| Folk | 4.65 |
| Jazz | 7.10 |
| Folk | 2.42 |

Hold on to that. Everything from here to the end of the course is a
variation on it:

* the `csv` module hands you exactly this shape (next notebook)
* a SQL query returns exactly this shape (session 6)
* a pandas DataFrame **is** this shape, with a much faster engine underneath
  (session 8)
