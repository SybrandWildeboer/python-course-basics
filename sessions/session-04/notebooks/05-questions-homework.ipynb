# Homework: two real questions

## Question 1: which genre has the highest average minutes per play?

Not the highest **total**: the highest **average**. A genre with 500 short
plays should not beat a genre with 100 long ones.

**The hint, because this is the interesting bit:** you need **two**
dictionaries. One counting the plays per genre, one adding up the minutes per
genre. Then divide one by the other, genre by genre.

That insight, "totals and counts kept side by side", is exactly what
`GROUP BY` does for you in session 5. Doing it by hand once makes SQL feel
obvious rather than magic.

## Question 2: which artist was played most in 2025 only?

The dates are text, like `"2025-04-17"`, and text has a `.startswith()`
method.

Print the **top five** artists for 2025 with their play counts, not just the
winner.

```python
import csv
import os
from pathlib import Path

here = Path.cwd()
while not (here / "data" / "clean" / "plays.csv").exists() and here != here.parent:
    here = here.parent
os.chdir(here)


def load_rows(path):
    """Return the CSV at path as a list of dictionaries."""
    with open(path, encoding="utf-8") as f:
        return list(csv.DictReader(f))


rows = load_rows("data/clean/plays.csv")
len(rows)
print(rows[:2])
```

## Question 1

```python
# TODO: two dictionaries, then divide
def function(rows):

    minutes = {}
    plays = {}
    averages = {}

    for x in rows:
        k = x["genre"]
        plays[k] = plays.get(k, 0) + 1
        v = float(x["minutes_played"])
        minutes[k] = minutes.get(k, 0) + v

    for y in minutes:
        averages[y] = round(minutes.get(y,0) / plays[y], 2)

    return minutes, plays, averages # this governs the order in which the dictionaries will be returned

a, b, c = function(rows) # here I assigned some letters to the dictionaries from the function above; recommended to have the same names.
print(a)
print(b)
print(c)

def top(counts, n):
    return sorted(counts.items(), key=lambda pair: pair[1], reverse=True)[:n]

best = top(counts=c, n=1)
print(f"The highest average per genre is {best}")

# alternitavely, I could print the dictionaries inside the function before return. and then I just need to call the function "function(rows)"

# vs. 
# result = function(rows)
# print(result)
```
