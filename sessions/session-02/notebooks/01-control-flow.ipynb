## 1. Comparisons give you True or False

```python
print(5 > 3)
print(5 < 3)
print(5 == 5)         # equal?
print(5 != 5)         # not equal?
print(5 >= 5)
print("a" == "A")     # capitals matter
```

**One `=` or two?**

* `=` puts a value into a variable
* `==` asks whether two things are equal

Mixing them up is the most common slip in this session, and Python usually
tells you off for it in a helpful way.

A value that is either `True` or `False` is called a **boolean**, after
George Boole. That is the whole reason for the odd name.

```python
age = 41
print(age > 18)
```

## 2. `if`: run this bit only sometimes

```python
temperature = 31

if temperature > 25:
    print("Warm out")
    print("Take water")

print("This always runs")
```

Two things carry the meaning there:

* the **colon** at the end of the `if` line
* the **indentation** of the lines under it

Indented lines belong to the `if`. The unindented line does not. In most
languages indentation is decoration; in Python it is grammar.

Four spaces is the convention, and pressing Tab gives you them.

Change `temperature` to 12 and run the cell again. Two lines disappear.

## 3. `elif` and `else`: the other branches

```python
score = 72

if score >= 90:
    print("Excellent")
elif score >= 70:
    print("Good")
elif score >= 50:
    print("Pass")
else:
    print("Not yet")
```

Only **one** branch runs. Python checks top to bottom and stops at the first
`True`. A score of 95 prints "Excellent" and never looks at the rest.

Which means **order matters**. Move the `>= 50` test to the top and run it
with `score = 95`: everything above 50 becomes a "Pass", including 95. The
program still runs perfectly and the answer is wrong. Those are the
expensive bugs.

Try it in the cell below.

```python
score = 95

if score >= 50:
    print("Pass")            # this now catches everything
elif score >= 90:
    print("Excellent")       # and this can never happen
```

## 4. Combining conditions: `and`, `or`, `not`

| Written | True when |
|---|---|
| `A and B` | both are true |
| `A or B` | at least one is true |
| `not A` | A is false |

```python
age = 41
member = True

if age > 18 and member:
    print("Full access")

if age < 12 or age > 65:
    print("Reduced price")

if not member:
    print("Please sign up")
```

## 5. `for` and `range`: do this once per item

```python
for i in range(5):
    print(i)
```

Five numbers, **starting at zero**. That will feel arbitrary until session 4,
where list positions also start at zero, and then it will click.

```python
for i in range(1, 6):
    print(i)              # 1 2 3 4 5, and NOT 6
```

`range(1, 6)` stops **before** 6. The end is never included. It feels wrong
for about a week and then it feels normal.

```python
for i in range(0, 20, 5):
    print(i)              # every fifth number
```

```python
for letter in "cat":
    print(letter)         # strings can be looped over too
```

The variable after `for` is yours to name. `i` is traditional for a counter,
but `for minute in ...` is much better when it means something.

## 6. The pattern you will use forever: accumulate

```python
total = 0                    # 1. set up BEFORE the loop

for i in range(1, 11):
    total += i               # 2. update INSIDE the loop

print(total)                 # 3. use AFTER the loop
```

`total += i` means "add `i` to whatever is already in `total`". It is short
for `total = total + i`.

**The classic bug** is putting `total = 0` inside the loop. Then it resets
every time round and the answer is always just the last number. Run this and
watch it happen:

```python
for i in range(1, 11):
    total = 0                # wrong place
    total += i

print(total)                 # 10, not 55
```

## 7. `while`: keep going until something changes

```python
count = 3

while count > 0:
    print(count)
    count -= 1          # THIS is the line that ends it

print("Go")
```

**Which loop when?**

* `for` when you know how many, or you have a collection of things
* `while` when you are waiting for a condition, like "until the guess is
  right"

**Infinite loops.** Forget the `count -= 1` and it runs forever. In a
notebook you stop a runaway cell with the **interrupt** button in the
toolbar (the square stop icon), or **Kernel, Interrupt**. In the terminal it
is **Ctrl + C**.

Worth doing on purpose once, so that the first time it happens for real you
shrug instead of panicking. Uncomment, run, then interrupt it.

## 8. `break` and `continue`

```python
for i in range(1, 100):
    if i * i > 50:
        print(f"{i} squared passes 50")
        break             # found it, stop looping
```

```python
for i in range(1, 11):
    if i % 2 == 0:
        continue          # skip the even ones
    print(i)
```

`i % 2 == 0` is the standard way to ask "is this even?". `%` gives the
remainder after dividing.

Both `break` and `continue` are conveniences. Anything you can do with them
you can do with a cleverer condition, but they usually read better.

## 9. When logic gets deep, flatten it

These two do the same thing. The second one you can read top to bottom like
a list of rules.

```python
logged_in, banned, credits = True, False, 0

# harder to follow
if logged_in:
    if not banned:
        if credits > 0:
            print("Welcome")
        else:
            print("Out of credits")
    else:
        print("Banned")
else:
    print("Please log in")
```

```python
# easier to follow: deal with the exceptions first
if not logged_in:
    print("Please log in")
elif banned:
    print("Banned")
elif credits == 0:
    print("Out of credits")
else:
    print("Welcome")
```

When you find yourself three levels deep, that is the nudge to restructure.
