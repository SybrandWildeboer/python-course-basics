# Homework: statistics by hand, worked solution

```python
numbers = [12, 7, 41, 3, 28, 19, 7, 55, 2, 33]
```

## The total

Start at zero, add each number in turn.

```python
total = 0
for number in numbers:
    total += number

print(total)
```

## The average

The total divided by how many there are.

```python
average = total / len(numbers)
print(f"{average:.1f}")
```

## The highest

Start by assuming the **first number** is the winner, then replace it
whenever we meet something bigger.

```python
highest = numbers[0]
for number in numbers:
    if number > highest:
        highest = number

print(highest)
```

## The lowest

The same idea with the comparison flipped.

```python
lowest = numbers[0]
for number in numbers:
    if number < lowest:
        lowest = number

print(lowest)
```

## All four together

```python
print(f"Count:   {len(numbers)}")
print(f"Total:   {total}")
print(f"Average: {average:.1f}")
print(f"Highest: {highest}")
print(f"Lowest:  {lowest}")
```

## Why `highest = numbers[0]` and not `highest = 0`

Starting at zero works for this list, purely by luck, because every number
in it is positive.

Here is the same code on a list of negative numbers. Nothing in the list is
bigger than 0, so the `if` never fires and the answer comes out as 0, which
is not even one of the numbers.

```python
negatives = [-5, -2, -9]

highest_wrong = 0
for number in negatives:
    if number > highest_wrong:
        highest_wrong = number

print("starting at zero:      ", highest_wrong)      # 0, and wrong

highest_right = negatives[0]
for number in negatives:
    if number > highest_right:
        highest_right = number

print("starting at the first: ", highest_right)      # -2, and correct
```

The bug does not crash. It quietly lies. Those are the expensive ones.

Starting with the first item of the list cannot have that problem: the
answer is always a number that was actually in the data.

In real code you would write `max(numbers)` and `min(numbers)`, and they do
exactly this internally. Writing it once by hand is the point of the
exercise.

```python
print(max(numbers), min(numbers), sum(numbers))
```

## The stretch goal

```python
above = 0
for number in numbers:
    if number > average:
        above += 1

print(f"{above} of {len(numbers)} numbers are above the average")
```

## One edge case worth knowing about

`numbers[0]` fails on an **empty** list with an `IndexError`, and
`total / len(numbers)` fails with a `ZeroDivisionError`.

Neither matters for a list you typed yourself. Both matter the moment the
list comes from a file that turned out to have no rows in it, which happens
in session 6. Handling it properly needs `try` and `except`, and that is on
the map for session 8.
