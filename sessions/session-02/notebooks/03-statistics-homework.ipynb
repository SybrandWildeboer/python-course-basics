Print the **total**, the **average**, the **highest** and the **lowest** of
the list below.

```python
numbers = [12, 7, 41, 3, 28, 19, 7, 55, 2, 33]
```

## The total

Set up a variable before the loop, update it inside, use it after.

```python
# TODO: the total
total = 0
for i in numbers:
    total += i
print (total)
```

## The highest

**This is the interesting one.** What do you set your "highest so far"
variable to *before* the loop starts?

Zero is the obvious answer and it is wrong. Work out why by trying your
version with `numbers = [-5, -2, -9]`.

Sitting with that for ten minutes is worth more than being told the answer.

```python
# TODO: the highest
highest = numbers [0]
for i in numbers:
    if i > highest: 
        highest = i
        # i += 1 - this is done automatically by for i in numbers; no need to add this
    #  i += 1 - same

print (highest)
```

## The lowest

```python
# TODO: the lowest
lowest = numbers [0]
for i in numbers:
    if i < lowest: lowest = i

print (lowest)
```
