# Errors, on purpose

Errors are not the program telling you off. They are the most useful message
you will get all day.

## The habit: read the last line first

```
Traceback (most recent call last):
  File "sums.py", line 4, in <module>
    total = total + price
NameError: name 'price' is not defined
```

1. **Last line.** What kind of problem, and a short description.
   `name 'price' is not defined` means Python has never heard of `price`.
2. **The line number.** `line 4`. That is where it gave up, and nearly
   always where the fix goes.
3. **The middle.** The path through your program to get there. One line long
   today; in a few sessions it will not be, and then it earns its keep.

Everybody reads that wall of text from the top and gives up. Go straight to
the bottom. This one habit will save you more time than anything else on
this course.

## Now break things

Each cell below is broken in a different way, and commented out so the
notebook runs clean. For each one:

1. delete the `#` marks
2. run the cell
3. **read the last line out loud, in your own words**
4. fix it, and see it work
5. move on

You are building a lookup table in your head. In three weeks you want to
recognise these in half a second.

### SyntaxError: missing colon

Python cannot even read the line.

````python
# if 5 > 3 print("hi")
````

### IndentationError

The body of an `if` has to be indented.

````python
# if 5 > 3:
# print("hi")
````

### NameError

A name Python has never seen. Almost always a typo, or a variable you meant
to create earlier. Check your capitals too.

````python
# total = 10
# print(totl)
````

### TypeError

Text and number mixed together. Convert one of them.

````python
# print("5" + 5)
````

### ZeroDivisionError

Exactly what it says. Watch for this in session 9, when you divide by a
count that happens to be zero.

````python
# people = 0
# print(100 / people)
````

## One extra error that only notebooks give you

Run the cell below, then run it a second time. It works the second time and
not the first.

```python
print(mystery)
mystery = 5
```

A notebook remembers everything you have run, so a cell can pass because of
something you ran ten minutes ago and have since deleted. That is the
**run-order trap** from the first notebook.

When results stop making sense: **Restart and Run All**. If the notebook
cannot run cleanly from top to bottom, it is not finished, however good the
output looks on screen.

## Your collection so far

| Error | Means |
|---|---|
| `SyntaxError` | Python cannot read the line. Missing colon, quote or bracket |
| `IndentationError` | Spacing is wrong at the start of a line |
| `NameError` | A name Python has never seen. Typo, or never assigned |
| `TypeError` | Right idea, wrong types. Text where a number belongs |
| `ZeroDivisionError` | Dividing by zero, often by a count that came out empty |
| `IndexError` | Asked for a list position that does not exist (session 4) |
| `KeyError` | Asked for a dictionary key that is not there (session 4) |
| `FileNotFoundError` | The file is fine, Python is looking elsewhere (session 4) |
