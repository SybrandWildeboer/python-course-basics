# Live exercise: exploring the listening log, worked solution

````python
import csv
import os
from pathlib import Path

here = Path.cwd()
while not (here / "data" / "clean" / "plays.csv").exists() and here != here.parent:
    here = here.parent
os.chdir(here)

PLAYS = "data/clean/plays.csv"
````

## Loading

````python
def load_rows(path):
    """Return the CSV at path as a list of dictionaries, one per row."""
    with open(path, encoding="utf-8") as f:
        return list(csv.DictReader(f))


rows = load_rows(PLAYS)
len(rows)
````

## 1. How many plays in total

`len()` on the list of rows. No loop needed.

````python
print(f"{len(rows)} plays in the log")
````

## 2. Plays per device

The column name is a **parameter**, so one function answers "per device",
"per genre" and "per country".

````python
def count_by(rows, column):
    """Return a dictionary of how many rows fall into each value of column."""
    counts = {}
    for row in rows:
        value = row[column]
        counts[value] = counts.get(value, 0) + 1
    return counts


def show_counts(title, counts):
    """Print a dictionary of counts as an aligned table, biggest first."""
    print(title)
    for key, number in sorted(counts.items(), key=lambda pair: pair[1], reverse=True):
        print(f"  {key:12} {number:>5}")


show_counts("plays per device", count_by(rows, "device"))
````

## 3. Total minutes

`float()` on every value, because everything from a CSV is text.

````python
def total_minutes(rows):
    """Return the total of the minutes_played column as a number."""
    total = 0.0
    for row in rows:
        total += float(row["minutes_played"])
    return total


print(f"{total_minutes(rows):,.0f} minutes listened in total")
````

## 4. Skipped plays, and the percentage

The column holds the text `"1"` or `"0"`, so compare against a string.

````python
def count_skipped(rows):
    """Return how many rows were skipped."""
    skipped = 0
    for row in rows:
        if row["skipped"] == "1":
            skipped += 1
    return skipped


skipped = count_skipped(rows)
print(f"{skipped} plays skipped, {skipped / len(rows) * 100:.1f}% of the total")
````

## The payoff

`count_by` was written for the device question and answers three more
without a single change, because the column is a parameter rather than being
baked in.

````python
show_counts("plays per genre", count_by(rows, "genre"))
````

````python
show_counts("plays per country", count_by(rows, "country"))
````

If you wrote three near-identical functions instead, that is a perfectly
reasonable first draft. Noticing that they differ by one word, and turning
that word into a parameter, is the refactor worth making.
