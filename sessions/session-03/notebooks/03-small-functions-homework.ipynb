```python
# TODO: def is_even(n):
def is_even(n):
    if n % 2 == 0: 
        return True
    else: return False

print(is_even(5))
```

```python
# TODO: def celsius_to_fahrenheit(c):
def celsius_to_fahrneheit(c):
    f = (c * 9 / 5 + 32)
    return f

print(celsius_to_fahrneheit(24))
```

```python
# TODO: def initials(full_name):
def initials(full_name):
    a = full_name[0]

    for i in range(len(full_name)): # ca sa citesc pozitia trebuie sa folosesc range(len(x)); for i in ... citeste fiecare caracter.
        if full_name[i]== " ":
            i += 1
            break
    b = full_name[i]
    print (f"{a}.{b}.")

initials("Emi Grosu")
```

```python
# TODO: def longest_word(sentence):
def longest_word(sentence):
    longest = ""
    current_word = ""

    for i in range(len(sentence)):
        if sentence[i] != " ":
            current_word += sentence[i]
        else:
            if len(current_word) > len(longest):
                longest = current_word

            current_word = ""

    if len(current_word) > len(longest):
        longest = current_word

    return longest

print (longest_word("ab cffe sdedsafasf"))
```
