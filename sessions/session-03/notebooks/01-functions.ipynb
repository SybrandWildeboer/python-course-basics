## 1. `def`: defining and calling

```python
def greet():
    print("Hello there")


greet()
greet()
```

Two separate moments there:

* **defining** the function teaches Python the recipe. Nothing happens yet.
* **calling** it, with the brackets, is what runs the recipe.

Without the brackets you are talking about the function rather than running
it, which Python will happily show you:

```python
print(greet)
```

## 2. Parameters: giving the function something to work with

```python
def greet_person(name):
    print(f"Hello {name}")


greet_person("Sam")
greet_person("Ana")
```

```python
def describe(name, age):
    print(f"{name} is {age}")


describe("Sam", 41)                # by position
describe(age=41, name="Sam")       # by name, so the order stops mattering
```

The **parameter** is the name in the `def` line. The **argument** is the
actual value you pass in. Nobody minds if you mix the words up.

# 3. `return` is not `print`

This is the important one. Slow down here.

Two functions that look like they do the same thing:

```python
def double_p(n):
    print(n * 2)          # shows a human


def double_r(n):
    return n * 2          # hands a value back to the program
```

Predict what each of the next four cells will show, before you run them.

```python
double_p(5)
```

```python
double_r(5)
```

The second one showed `10` too, but for a different reason: a notebook
displays the value of the last line of a cell. In a `.py` script,
`double_r(5)` on its own shows **nothing at all**.

The difference becomes obvious the moment you try to *use* the answer.

```python
result_p = double_p(5)
print("from the print version: ", result_p)

result_r = double_r(5)
print("from the return version:", result_r)
```

`double_p` gave back `None`. It showed you the answer and kept nothing.

Now the line that makes it click. Uncomment it and run.

```python
print(double_r(5) + 1)        # 11
# print(double_p(5) + 1)
# TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'
```

**print shows a human. return hands a value back to your program.**

A function that prints is a dead end: you can look at the answer but you
cannot do anything with it. A function that returns can be added up, sorted,
passed into another function, or written to a file.

The rule of thumb for the rest of the course:

> Functions that calculate should **return**. Only the outermost part of
> your program should **print**.

And because a returned value is a real value, you can feed one into another:

```python
print(double_r(double_r(3)))      # 12
```

## 4. No `return` means `None`

```python
def save_it(text):
    print(f"saving {text}")
    # no return line


x = save_it("hello")
print(x, type(x))
```

`None` is a real value that means "nothing here". It is not zero, not empty
text, and not an error.

Here is the bug to remember: an `if` with a `return` inside it and no other
path hands back `None` whenever the condition is false.

```python
def risky(n):
    if n > 0:
        return "positive"
    # falls off the end for anything else


print(risky(5))
print(risky(-5))        # None
```

Then some line far away fails with **NoneType** in the message. When you see
that word, look for a function that forgot to return.

## 5. `return` stops the function there and then

```python
def check(age):
    """Return a short verdict on an age."""
    if age < 0:
        return "not a real age"
    if age < 18:
        return "too young"
    return "fine"


print(check(-3))
print(check(12))
print(check(41))
```

That shape is called a **guard clause**: deal with the odd cases first and
return immediately, so what is left at the bottom is the normal case,
unindented and easy to read. It is the flattening idea from last week,
applied to functions.

## 6. Default values

```python
def greeting_for(name, greeting="Hello"):
    return f"{greeting} {name}"


print(greeting_for("Sam"))
print(greeting_for("Sam", "Morning"))
print(greeting_for("Sam", greeting="Oi"))
```

A default makes a parameter optional: the function gets a sensible everyday
behaviour and stays adjustable.

Parameters with defaults have to come **after** the ones without.
`def f(a=1, b)` is a `SyntaxError`, and the message says so clearly.

## 7. Scope: what happens inside stays inside

```python
def calculate():
    subtotal = 100          # born here, gone when the function ends
    return subtotal * 1.21


print(calculate())
```

Uncomment the next line. `subtotal` does not exist out here.

```python
# print(subtotal)
# NameError: name 'subtotal' is not defined
```

That is a feature, not a restriction. It means you can use the name `total`
in ten different functions without them interfering with each other.

Names defined *outside* a function are visible inside it:

```python
VAT = 1.21


def calculate_again(amount):
    return amount * VAT


print(calculate_again(100))
```

**The rule of thumb:** pass what you need in as parameters, hand the answer
back with `return`. Do not reach out of a function for values.

## 8. Docstrings and names

A docstring is a string on the first line of a function. Your editor shows
it when you hover, and `help()` prints it.

```python
def average(numbers):
    """Return the mean of a list of numbers.

    Assumes the list is not empty.
    """
    total = 0
    for number in numbers:
        total += number
    return total / len(numbers)


print(average([1, 2, 3]))
help(average)
```

**Name it after what it gives you:**

* `average(numbers)`, not `do_calc(x)`
* verbs for actions: `load_data`, `save_chart`
* questions for true or false: `is_even`, `has_errors`

## 9. Functions calling other functions

This is the whole idea. Small functions that each do one thing, stacked into
bigger ones.

```python
def total(numbers):
    """Return the sum of a list of numbers."""
    running = 0
    for number in numbers:
        running += number
    return running


def mean(numbers):
    """Return the mean of a list of numbers."""
    return total(numbers) / len(numbers)      # reuse, do not repeat


def summarise(numbers):
    """Return one readable line describing a list of numbers."""
    return (f"{len(numbers)} values, "
            f"total {total(numbers)}, "
            f"average {mean(numbers):.1f}")


print(summarise([12, 7, 41, 3]))
```

`mean` does not know how to add up a list. It asks `total`. Fix a bug in
`total` and both are fixed.

Notice that only the last line prints. Everything above it returns. That is
the shape to aim for: **calculate quietly, print once, at the edge.**

## 10. Checking a function on cases you can work out in your head

```python
def is_even(n):
    """Return True if n divides by 2 exactly."""
    return n % 2 == 0


print(is_even(4), is_even(7), is_even(0))
```

The tidier way to write those checks down. `assert something_true` does
nothing at all; if it is false you get an `AssertionError` on that line.
Silence means everything passed.

```python
assert is_even(4) is True
assert is_even(7) is False
assert is_even(0) is True        # the awkward case, checked on purpose
print("all checks passed")
```

**Pick the awkward cases:** zero, one, negatives, an empty list. That is
where the bugs live. Your `highest` function last week was correct for the
list you tried and wrong for `[-5, -2, -9]`.

This is the seed of proper testing, which is on the map in session 12 under
`pytest`.

## The five function mistakes, and what they look like

| Mistake | What you see | Fix |
|---|---|---|
| Defined it, never called it | Nothing happens, no error at all | Add the call |
| Called it without brackets | Nothing, or `<function greet at 0x...>` | `greet()`, not `greet` |
| Printed instead of returned | `TypeError` mentioning `NoneType` | `return x`, not `print(x)` |
| A `return` inside an `if`, no other path | `None` for some inputs, no complaint | Make every route return |
| Using an inside variable outside | `NameError` | Return it, store what comes back |

**Three of those five produce no error message at all.** When a function
seems to do nothing, start at the top of that table.
