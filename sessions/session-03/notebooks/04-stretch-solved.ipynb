## The four functions from the live exercise

```python
def total(numbers):
    """Return the sum of a list of numbers."""
    running = 0
    for number in numbers:
        running += number
    return running


def average(numbers):
    """Return the mean of a list of numbers."""
    return total(numbers) / len(numbers)


def highest(numbers):
    """Return the largest value in a list of numbers."""
    best = numbers[0]
    for number in numbers:
        if number > best:
            best = number
    return best


def lowest(numbers):
    """Return the smallest value in a list of numbers."""
    worst = numbers[0]
    for number in numbers:
        if number < worst:
            worst = number
    return worst


NUMBERS = [12, 7, 41, 3, 28, 19, 7, 55, 2, 33]
```

## Stretch 1: `spread`

```python
def spread(numbers):
    """Return the distance between the largest and smallest value."""
    return highest(numbers) - lowest(numbers)


assert spread([3, 10, 7]) == 7
assert spread([5]) == 0
assert spread([-5, -2, -9]) == 7
print(spread(NUMBERS))
```

One line, no loop. The loops already exist inside `highest` and `lowest`,
so writing a third one would be repeating yourself.

## Stretch 2: `count_above`

```python
def count_above(numbers, threshold):
    """Return how many values are strictly greater than threshold."""
    count = 0
    for number in numbers:
        if number > threshold:
            count += 1
    return count


assert count_above([1, 5, 10], 4) == 2
assert count_above([1, 5, 10], 10) == 0      # "above" means strictly above
assert count_above([1, 5, 10], 100) == 0

print(f"{count_above(NUMBERS, average(NUMBERS))} of {len(NUMBERS)} values are above average")
```

`count_above(NUMBERS, average(NUMBERS))` is a function call used as an
argument to another function call. Python works from the inside out: first
`average(NUMBERS)` becomes `20.7`, then `count_above(NUMBERS, 20.7)` runs.

That only works because `average` **returns**. If it printed, the inner
call would hand `None` to `count_above`, and `number > None` would crash.

## Stretch 3: `median`

```python
def median(numbers):
    """Return the middle value of a list, or the mean of the two middle values."""
    ordered = sorted(numbers)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:
        return ordered[middle]
    return (ordered[middle - 1] + ordered[middle]) / 2


assert median([3, 1, 2]) == 2
assert median([4, 1, 3, 2]) == 2.5
assert median([7]) == 7

before = list(NUMBERS)
print(median(NUMBERS))
assert NUMBERS == before      # sorted() made a copy; the original is untouched
```

Walk through the even case with `[4, 1, 3, 2]`:

* `ordered` is `[1, 2, 3, 4]`, length 4
* `middle` is `4 // 2`, which is `2`
* the two middle values are at positions `1` and `2`: that is `2` and `3`
* their mean is `2.5`

**Mean versus median.** For `NUMBERS` the mean is 20.7 and the median is
15.5. The mean is pulled up by the 55. That gap is why house prices and
salaries are usually reported as medians: one huge value does not move the
median at all.

Remember `ordered[middle - 1]` for the next session: lists are counted from
0, which is why the 2nd item is at position 1.

## Stretch 4: `format_duration`

```python
def format_duration(seconds):
    """Return a number of seconds as minutes:seconds, like '3:07'."""
    minutes = seconds // 60
    rest = seconds % 60
    return f"{minutes}:{rest:02d}"


assert format_duration(187) == "3:07"
assert format_duration(59) == "0:59"
assert format_duration(60) == "1:00"
assert format_duration(0) == "0:00"
```

The hours version. The check for hours comes first, and returns early, so
the rest of the function only ever sees durations under an hour.

```python
def format_duration(seconds):
    """Return seconds as m:ss, or h:mm:ss for an hour or more."""
    if seconds >= 3600:
        hours = seconds // 3600
        minutes = (seconds % 3600) // 60
        rest = seconds % 60
        return f"{hours}:{minutes:02d}:{rest:02d}"
    return f"{seconds // 60}:{seconds % 60:02d}"


assert format_duration(187) == "3:07"
assert format_duration(3600) == "1:00:00"
assert format_duration(3725) == "1:02:05"
assert format_duration(3599) == "59:59"       # just under an hour
```

Defining `format_duration` a second time replaced the first one. Python
keeps whichever definition ran last, so the order you run cells in matters.

## Stretch 5: what should `average([])` do?

```python
def safe_average(numbers):
    """Return the mean of a list, or None if the list is empty."""
    if len(numbers) == 0:
        return None
    return average(numbers)


assert safe_average([]) is None
assert safe_average([2, 4]) == 3

result = safe_average([])
if result is None:
    print("no plays yet, so no average")
else:
    print(f"average: {result:.1f}")
```

**Why not return 0?** Because 0 is a real average. An empty list and a list
of zeros would then look identical, and a total like "average rating 0"
would quietly lie in a report. `None` means "there is no answer", and
anyone using the function is forced to deal with that.

**Why not let it crash?** Sometimes that is the right call. A crash is
loud and honest. The choice depends on whether an empty list is a normal
situation (a new user with no plays yet: return `None`) or a sign
something went wrong (a file that should never be empty: crash).

Either way, *decide* and write it in the docstring. The bad version is the
one nobody thought about.

`None` is checked with `is`, not `==`. Both work here, but `is None` is
what you will see everywhere, so it is worth getting used to.

## Stretch 6: `describe_track`

```python
def describe_track(title, artist, seconds):
    """Return one line describing a track, like 'Hey Jude by The Beatles (7:11)'."""
    return f"{title} by {artist} ({format_duration(seconds)})"


assert describe_track("Hey Jude", "The Beatles", 431) == "Hey Jude by The Beatles (7:11)"
print(describe_track("Bohemian Rhapsody", "Queen", 354))
print(describe_track("Echoes", "Pink Floyd", 1412))
```

Four functions deep now: `describe_track` calls `format_duration`, which
uses `//` and `%` from session 1. Each piece is small enough to check on its
own, and that is the whole idea. In session 4 you will run this over every
track in the real dataset.
