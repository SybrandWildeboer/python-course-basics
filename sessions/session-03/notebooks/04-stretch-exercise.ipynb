## Stretch 3: `median`

The median is the middle value once the list is in order.

* `[3, 1, 2]` in order is `[1, 2, 3]`, so the median is `2`
* `[4, 1, 3, 2]` in order is `[1, 2, 3, 4]`. There is no single middle, so
  the median is the mean of the two middle values: `2.5`
