# SQL and pandas, side by side

Four queries from sessions 6 and 7, each written twice: once in SQL, run
through `pd.read_sql`, and once in pandas. After each pair there is an
`assert` that the two answers are the same. Two independent routes to one
number is the strongest check you can give yourself.

The fourth query is fan-out, and it is the reason this notebook exists.

```python
import os
import sqlite3
from pathlib import Path

import pandas as pd

here = Path.cwd()
while not (here / "data" / "music.db").exists() and here != here.parent:
    here = here.parent
os.chdir(here)

con = sqlite3.connect("data/music.db")


def sql(query):
    """Run a query against music.db and return the result as a DataFrame."""
    return pd.read_sql(query, con)


plays = sql("SELECT * FROM plays")
artists = sql("SELECT * FROM artists")
awards = sql("SELECT * FROM awards")

print(plays.shape, artists.shape, awards.shape)
```

All three tables come from the database, so the pandas versions start from
exactly what the SQL sees: `plays` has `artist_id`, not names.

---

## Query 1: filter and sort

Session 6, rungs 4 and 5 of the ladder: the plays on the speaker in 2025,
longest first.

```python
in_sql = sql("""
SELECT played_at, track_name, minutes_played
FROM plays
WHERE device = 'speaker'
  AND played_at BETWEEN '2025-01-01' AND '2025-12-31'
ORDER BY minutes_played DESC
""")
in_sql.head()
```

```python
speaker_2025 = plays[(plays["device"] == "speaker") & (plays["played_at"] >= "2025-01-01")]
in_pandas = speaker_2025.sort_values("minutes_played", ascending=False)
in_pandas = in_pandas[["played_at", "track_name", "minutes_played"]]
in_pandas.head()
```

```python
assert len(in_sql) == len(in_pandas) == 174
assert list(in_sql["minutes_played"]) == list(in_pandas["minutes_played"])
print("same 174 rows, in the same order")
```

Notice the order of the pandas lines: filter, then sort, then pick the
columns. SQL makes you write `SELECT` first even though it happens last.
pandas has you write things in the order they happen.

The dates need only `>=` here, because the data stops at the end of 2025.
SQL's `BETWEEN` has a pandas twin, `.between("2025-01-01", "2025-12-31")`,
if you want the end to be explicit.

## Query 2: group, then filter the groups

Session 6, rung 10: the devices with more than 300 plays.

```python
in_sql = sql("""
SELECT device, COUNT(*) AS plays
FROM plays
GROUP BY device
HAVING COUNT(*) > 300
ORDER BY plays DESC
""")
in_sql
```

```python
per_device = plays["device"].value_counts()
in_pandas = per_device[per_device > 300]
in_pandas
```

```python
assert list(in_sql["device"]) == list(in_pandas.index) == ["phone", "laptop", "speaker"]
assert list(in_sql["plays"]) == list(in_pandas) == [1039, 524, 336]
print("same three devices, same counts")
```

There is no `HAVING` in pandas, and nothing is missing. `HAVING` exists in
SQL because the database needs telling whether a filter applies before or
after the grouping. In pandas the filter is just on the line after the
count, so it obviously applies to the counts.

This is the place where pandas is easier to think in. The awkward
"you cannot use `COUNT(*)` in a `WHERE`" error from session 6 cannot happen.

## Query 3: every artist, including the ones never played

Session 7, slide 11: a `LEFT JOIN`, and `COUNT(p.play_id)` rather than
`COUNT(*)`.

```python
in_sql = sql("""
SELECT a.artist_name, a.genre, COUNT(p.play_id) AS plays
FROM artists a
LEFT JOIN plays p USING (artist_id)
GROUP BY a.artist_id
ORDER BY plays
LIMIT 4
""")
in_sql
```

```python
everyone = artists.merge(plays, on="artist_id", how="left")
print(len(everyone), "rows after the left merge")

per_artist = everyone.groupby(["artist_name", "genre"]).agg(plays=("play_id", "count"))
in_pandas = per_artist.sort_values("plays").head(4)
in_pandas
```

`how="left"` is `LEFT JOIN`. `groupby` with **two** columns groups by the
pair, which is how the genre comes along for the ride, the way
`GROUP BY a.artist_id` let SQL show the name and genre.

Now the trap, which has survived the translation intact. `count` counts the
values that are not missing, like `COUNT(p.play_id)`. `size` counts rows,
like `COUNT(*)`.

```python
wrong = everyone.groupby("artist_name").size().sort_values().head(4)
wrong
```

```python
right = everyone.groupby("artist_name")["play_id"].count()

assert wrong["Iron Fernway"] == 1        # the COUNT(*) bug, in pandas
assert right["Iron Fernway"] == 0
assert right["Gravel Choir"] == 0
assert set(in_sql["artist_name"].head(2)) == {"Iron Fernway", "Gravel Choir"}
print("size says 1, count says 0, the truth is 0")
```

Same bug, same fix, same reason: after a left join, an artist with no plays
still has one row, full of `NaN`. `size` counts that row. `count` skips the
`NaN` and says 0. Change one word and "never played" becomes "played once".

## Query 4: fan-out

Session 7's wrong total. First the true total, both ways.

```python
true_sql = sql("SELECT ROUND(SUM(minutes_played), 1) AS total FROM plays")["total"][0]
true_pandas = round(plays["minutes_played"].sum(), 1)

assert true_sql == true_pandas == 8980.4
print(true_pandas)
```

Now join `awards` on, and add up the minutes again.

```python
fan_sql = sql("""
SELECT ROUND(SUM(p.minutes_played), 1) AS total
FROM plays p
JOIN awards w ON w.artist_id = p.artist_id
""")["total"][0]

with_awards = plays.merge(awards, on="artist_id")
fan_pandas = round(with_awards["minutes_played"].sum(), 1)

assert fan_sql == fan_pandas == 9732.1
print(fan_pandas)
```

8,980.4 became 9,732.1 in both languages. pandas did exactly what SQL did,
without a word of complaint, because it is exactly the same operation.

And the same diagnosis: count the rows.

```python
print(len(plays), "rows before")
print(len(with_awards), "rows after")
```

```python
fjord = plays[plays["artist_id"] == 35]
fjord_awards = awards[awards["artist_id"] == 35]
fjord_after = with_awards[with_awards["artist_id"] == 35]

print(len(fjord), "plays,", len(fjord_awards), "awards,", len(fjord_after), "rows after the merge")
print(round(fjord["minutes_played"].sum(), 1), "minutes really,",
      round(fjord_after["minutes_played"].sum(), 1), "after the merge")
```

Fjord & Flint: 237 plays, 3 awards, 711 rows. Every play met every award, so
every minute was counted three times. That is fan-out, and it is called the
same thing in pandas as everywhere else.

### The pandas versions of the fake fix

In SQL the tempting fix was `SUM(DISTINCT)`. In pandas the tempting fix is
`drop_duplicates()`, and it fails in a different way:

```python
print(len(with_awards.drop_duplicates()), "rows after drop_duplicates()")
```

It removes nothing at all. The 711 Fjord & Flint rows are not duplicates:
each one carries a different award, so no two rows are identical. And
de-duplicating on the minutes column alone is `SUM(DISTINCT)` all over
again:

```python
print(round(fjord_after["minutes_played"].drop_duplicates().sum(), 1))
```

680.6, the same wrong answer as session 7, for the same reason: two real
plays that both lasted 4.65 minutes count once. Do not reach for the switch
that makes the number look right.

### The real fix: aggregate first, then merge

In session 7 this was a CTE. In pandas, a CTE is just **a variable with a
good name**: reduce each table to one row per artist, and only then merge.

```python
per_artist = plays.groupby("artist_id").agg(
    plays=("play_id", "count"),
    minutes=("minutes_played", "sum"),
).reset_index()

awards_per_artist = awards.groupby("artist_id").size().rename("awards").reset_index()

summary = (
    per_artist
    .merge(artists[["artist_id", "artist_name"]], on="artist_id")
    .merge(awards_per_artist, on="artist_id", how="left")
)
summary["awards"] = summary["awards"].fillna(0).astype(int)

summary.sort_values("minutes", ascending=False).head(3).round(1)
```

Two details:

* `.reset_index()` turns `artist_id` from the index back into an ordinary
  column, so the next `merge` can use it with `on=`.
* `fillna(0)` is `COALESCE(x, 0)`: artists with no awards come back from the
  left merge as `NaN`, and here a missing count really does mean zero.
  `astype(int)` turns 2.0 back into 2, because a column holding `NaN` is
  always stored as decimals.

Now the verification, which matters as much as the fix:

```python
fixed_sql = sql("""
WITH per_artist AS (
    SELECT artist_id, COUNT(*) AS plays, SUM(minutes_played) AS minutes
    FROM plays
    GROUP BY artist_id
)
SELECT a.artist_name, ROUND(s.minutes, 1) AS minutes, COUNT(w.award_id) AS awards
FROM per_artist s
JOIN artists a USING (artist_id)
LEFT JOIN awards w USING (artist_id)
GROUP BY a.artist_id
ORDER BY s.minutes DESC
""")

fjord_row = summary[summary["artist_name"] == "Fjord & Flint"].iloc[0]

assert round(fjord_row["minutes"], 1) == 846.2      # the standalone number, exactly
assert fjord_row["awards"] == 3
assert round(summary["minutes"].sum(), 1) == 8980.4  # and nothing inflated overall
assert list(fixed_sql["artist_name"].head(3)) == ["Glass Tram", "DJ Kompas", "Fjord & Flint"]
print("846.2 and 3 awards, matching the CTE and the standalone query")
```

`.iloc[0]` means "the first row, by position". The filter gives back a small
table with one row in it, and `.iloc[0]` takes that row out.

### The guard SQL does not have: `validate`

You know which way round a merge should go before you write it. Plays to
artists is **many to one**: many plays, one artist each. pandas lets you say
that, and it will refuse to run if it is not true.

```python
checked = plays.merge(artists, on="artist_id", validate="many_to_one")
print(len(checked), "rows, and the check passed")
```

Now say the same thing about awards, where it is false:

```python
try:
    plays.merge(awards, on="artist_id", validate="many_to_one")
except pd.errors.MergeError as error:
    print("MergeError:", error)
```

"Merge keys are not unique in right dataset; not a many-to-one merge." The
fan-out that SQL let through silently is now a loud error, before a single
wrong number exists.

The aggregate-first version passes its checks, because after the `groupby`
both sides really do have one row per artist:

```python
safe = per_artist.merge(awards_per_artist, on="artist_id", how="left", validate="one_to_one")
print(len(safe), "rows, one per played artist")
```

The options are `"one_to_one"`, `"one_to_many"`, `"many_to_one"` and
`"many_to_many"`. Writing `validate=` on every merge costs a few characters
and turns an assumption in your head into a check the computer makes for
you. Of everything in this notebook, it is the habit most worth taking away.

## When SQL, when pandas

Honestly, for a dataset this size, either. You have now seen every question
answered both ways. The real differences show up elsewhere:

**Reach for SQL when**

* the data already lives in a database, especially a big one. The database
  does the filtering and grouping where the data is, and sends back only the
  answer. Pulling a hundred million rows into pandas to count them is slow,
  and sometimes impossible.
* other people need to run it. SQL is the shared language: analysts,
  engineers and dashboards all speak it.

**Reach for pandas when**

* the data is a file: a CSV, a spreadsheet, a download.
* it needs cleaning first. Fixing text, dates and numbers row by row is far
  easier in Python. That is session 9.
* the next step is a chart, or a loop, or a function, or anything else that
  is Python. Also session 9, and session 10.
* you are exploring, one step at a time, keeping each result in a variable.

**And very often, both:** SQL to fetch and reduce, `pd.read_sql` to cross
over, pandas for the rest. Fan-out does not care which one you use. The
row-count check works in both, and `validate=` is one reason to do the join
on the pandas side when you can.
