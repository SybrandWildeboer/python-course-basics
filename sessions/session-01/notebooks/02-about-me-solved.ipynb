## The script version, and the one new idea in it

In `solutions/about_me.py` the first two lines are different, because the
script asks instead of being told:

```python
name = input("What is your name? ")
birth_year = int(input("What year were you born? "))
```

Read that second line from the inside out, the way you read brackets in
maths:

1. `input(...)` runs first, and collects text like `"1985"`
2. `int(...)` wraps the result, turning it into the number `1985`

The `int()` is not optional. `input()` **always** gives you text, even when
the person types digits, and subtracting text from a number fails:

```
TypeError: unsupported operand type(s) for -: 'int' and 'str'
```

That error will be one of your most frequent companions for the next two
weeks. It is worth being able to read it at a glance.
