# Session 1: first steps

This is a **notebook**. It is the main way you will work on this course.

A notebook is a list of *cells*. Some cells are text, like this one. Some are
code, like the one below. You run a code cell with **Shift + Enter**, and the
answer appears underneath it.

Try it now: click the cell below and press Shift + Enter.

```python
print("Hello from my first notebook")
```

## Why a notebook?

Because you can work in small pieces. Write two lines, run them, look at the
answer, write two more. That loop, **edit, run, look**, is how you will learn
everything on this course, and a notebook makes the loop as short as
possible.

A `.py` file runs top to bottom, all at once. That is the right shape for a
program somebody else will use, and we come back to it later today.

## One thing to know about cells

Cells share their variables. A variable you make in one cell is available in
every cell after it. Run these two in order.

```python
greeting = "Hello"
```

```python
print(greeting, "again")
```

## The one trap in notebooks

Cells remember what you ran, **not the order they appear on screen**. If you
scroll up, change a cell and run it, everything below is now out of date and
the notebook will happily disagree with itself.

The number in `[1]` beside a cell tells you the order things actually ran in.

When something makes no sense, use **Restart and Run All** from the toolbar.
It clears everything and runs the notebook from the top, in order. It is the
notebook equivalent of turning it off and on again, and it is right
surprisingly often.

# 2. Variables

A variable is a name for a value.

```python
name = "Sam"
age = 41

print(name)
print(age)
```

`=` does not mean what it means in mathematics. It means **put this value in
here**. So this line, which is nonsense in algebra, is perfectly ordinary:
work out the right-hand side, then store it under that name.

```python
age = age + 1
print(age)
```

**Naming rules and manners**

* letters, digits and underscores; no spaces, cannot start with a digit
* case matters: `Name` and `name` are two different variables
* use `lower_case_with_underscores`, and use real words:
  `total_minutes`, not `tm`

# 3. Four types

| Type | Is | Example |
|---|---|---|
| `str` | text, a "string" of characters | `"Sam"` |
| `int` | a whole number | `41` |
| `float` | a number with a decimal point | `4.65` |
| `bool` | true or false, nothing else | `True` |

```python
print(type("Sam"))
print(type(41))
print(type(4.65))
print(type(True))
```

Here is the single most common beginner confusion, in two lines. Predict the
answers before you run it.

```python
print(4 + 4)          # numbers add
print("4" + "4")      # text joins
```

`"41"` with quotes is **text** that happens to look like a number. `41`
without quotes is a **number**. Python treats them completely differently.

A couple more arithmetic details worth meeting now:

```python
print(10 / 4)         # 2.5   dividing ALWAYS gives a float
print(10 // 4)        # 2     whole-number division
print(10 % 4)         # 2     the remainder
```

# 4. f-strings

Putting values inside text. The `f` before the opening quote is the whole
trick.

```python
print("Hi " + name + ", you are " + str(age))   # the tedious way
print(f"Hi {name}, you are {age}")              # the good way
```

```python
print(f"Next year: {age + 1}")        # any expression fits in the braces
print(f"Shouted: {name.upper()}")
```

You can control the formatting too. This comes back in session 9 for chart
labels.

```python
minutes = 1234.5678
print(f"{minutes:.1f} minutes")       # one decimal place
print(f"{minutes:,.2f} minutes")      # thousands separator, two decimals
```

Forget the `f` and you get the literal braces. This is a useful bug to
recognise on sight, because you will write it this week.

```python
print("Hi {name}, you are {age}")
```

# 5. Converting between types

```python
print(int("41"))        # 41     text -> whole number
print(float("4.65"))    # 4.65   text -> decimal
print(str(41))          # "41"   number -> text
print(int(4.99))        # 4      chops, does NOT round
print(round(4.99))      # 5      rounds properly
```

`int(4.99)` is **4**, not 5. It throws away everything after the point. When
you want real rounding, say `round()`.

Conversion can also fail, loudly. Uncomment the line below and run it to see
the error, then put the `#` back.

```python
# int("forty one")
# ValueError: invalid literal for int() with base 10: 'forty one'
```

# 6. Comments

Everything after a `#` is ignored by Python. Comments are notes for humans.

```python
VAT = 1.21          # the trailing kind works too

# Good comments say WHY:
# prices in the export are net, the report needs gross
gross = 45.00 * VAT
print(f"{gross:.2f}")
```

The second use for a `#` is switching a line off without deleting it.
**Ctrl + /** (Cmd + / on a Mac) toggles it. That will be your main debugging
tool for the next few weeks.

# Notebooks and scripts: when to use which

One thing is missing from this notebook: `input()`, which asks the person
running the program a question.

It technically works in a notebook, but it is awkward: a little box appears
at the top of the window and the cell sits there waiting. Asking questions is
a job for a `.py` script, run from the terminal.

So, the division you will use all course:

| Use a notebook when | Use a `.py` script when |
|---|---|
| exploring, in small steps | somebody needs to *run* the thing |
| you want the answer next to the code | it asks questions, or takes arguments |
| you are learning something new | it should run start to finish, unattended |
| you will write it up with prose | it needs to be scheduled, or shared as a tool |

Both files for this session's exercise exist in the repo so you can see the
difference for yourself:

  including the `input()` parts
