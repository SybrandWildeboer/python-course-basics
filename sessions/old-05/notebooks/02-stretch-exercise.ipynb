## Set up the playground

The playground must **not** be inside your course folder. A repo inside a
repo confuses both of them. So go up one level first:

```bash
cd ..
mkdir git-playground
cd git-playground
git init
code .
```

`code .` opens the folder in VS Code. (If your terminal does not know
`code`, open the folder from VS Code's File menu instead.) Create a file called `songs.txt`
with exactly this in it, and save:

```
Favourite songs

Hey Jude
Bohemian Rhapsody
```

Then make the first commit:

```bash
git add songs.txt
git commit -m "Add a list of songs"
```

From here on, "edit `songs.txt`" means edit it in VS Code and save.

## Stretch 1: read the history like a detective

Back in your **real** course repo. Nothing in this stretch changes
anything; it only reads.

`git log` takes options that turn it from a wall of text into something

1. Which commits did **you** write, as opposed to the ones that came with

2. Which commits changed **one particular file**? Pick any file of yours.

4. Find a commit by a word in its message. (Look up `--grep`.)
5. Run `git log --oneline --graph --all -15`. What are the lines and

## Stretch 2: fix the last commit

In the playground. Add the line `Echoes` to the end of `songs.txt`, then
commit it with a typo on purpose:

```bash
git add songs.txt
git commit -m "Add Echos"
git log --oneline
```

**New tool:** `git commit --amend` replaces the most recent commit with a
new one. Use it to fix the message to `Add Echoes`, then look at
`git log --oneline` again. Write down the commit's short code (the hash)
before and after. What happened to it?

**Harder:** amend can also add a file you forgot. Add `Wish You Were Here`
to `songs.txt` and commit it. Now add `Hotel California` too, and get it
into that **same** commit without changing the message. (Look up
`--no-edit`.) Check with `git show --stat HEAD`.

**Why question:** amending is safe here because the commit only exists on
your laptop. Why would amending a commit you have **already pushed** cause
trouble? Think about what you just saw happen to the hash.

*Your answer here.*

## Stretch 3: `.gitignore`, and the trap in it

In the playground. Create a file `scratch.txt` and a folder `private`
with a file `secret.txt` inside it. Run `git status`.

Now create a file called `.gitignore` (the dot matters) that makes git
ignore both of them. Run `git status` again until neither appears.

**New tool:** `git check-ignore -v private/secret.txt` tells you
*which line* of which `.gitignore` is ignoring a file. Try it on
`songs.txt` too. What does it say when a file is **not** ignored?

Commit the `.gitignore`.

**The trap.** Create `debug.log`, then add and commit it, as if by mistake.
Now add `*.log` to `.gitignore`, change a word in `debug.log`, and run
`git status`. Is git ignoring it?

Work out how to make git stop tracking `debug.log` **without deleting the
file from your folder**. (Look up `git rm --cached`.) Commit the fix and
check that `debug.log` is still in VS Code's file list.

## Stretch 4: put work aside with `git stash`

In the playground. First make a branch with one commit on it:

```bash
git switch -c experiment
```

Add `Stairway to Heaven` to the end of `songs.txt`, commit it, and
`git switch main`.

Now, on `main`, start something new: add the line `Half-finished idea` to
the end of `songs.txt`, save, but **do not commit**. Halfway through,
you want to look at the `experiment` branch. Try:

```bash
git switch experiment
```

Read the message. Git refuses, and it tells you the two ways out.

**New tool:** `git stash` puts your uncommitted changes on a shelf and
gives you a clean folder. `git stash list` shows the shelf, and
`git stash pop` puts the changes back.

Stash your half-finished line, switch to `experiment` and look at the
file, switch back to `main`, and get your half-finished line back. Check
with `git status`. Then throw the line away with `git restore songs.txt`.

## Stretch 5: undo a commit with `git revert`

In the playground, on `main`. Make two commits, one bad and one good:

1. add `Baby Shark` to the end of `songs.txt` and commit it as
   `Add Baby Shark`,
2. create `README.md` with one sentence about what this folder is, and
   commit it as `Add a README`.

Now you regret Baby Shark, but you want to keep the README. Find the Baby
Shark commit's hash with `git log --oneline`, and undo **only that
commit**:

```bash
git revert --no-edit <the hash>
```

Run `git log --oneline` and open `songs.txt`. Is Baby Shark gone? Is the
Baby Shark commit gone from the history?

**Why question, the one to talk through:** if you search online for "undo
a git commit", you will find `git reset --hard`. It makes the commit
disappear from the history completely, which sounds tidier. **Do not run
it.** Instead, explain: why is adding an *opposite* commit the safer way
to undo, especially once your work is pushed to GitHub?

*Your answer here.*

## Stretch 6: a merge conflict, on purpose

Sooner or later two changes will touch the same line and git will not
know which one you want. Better to meet that here than in real work.

In the playground, on `main`:

```bash
git switch -c new-title
```

Change the first line of `songs.txt` to `Songs I love`, and commit. Then
`git switch main`, change the **same line** to `My favourite songs`, and
commit that too. Two branches, two different first lines. Now:

```bash
git merge new-title
```

1. Read the message, then run `git status`. What does it say?
2. Open `songs.txt`. What are the `<<<<<<<`, `=======` and `>>>>>>>` lines
   marking? Which part came from which branch?
3. The emergency exit: `git merge --abort`. Run it, check `git status`,
   then run `git merge new-title` again to get the conflict back.
4. Resolve it: edit the file so it says what you actually want (one of
   the two titles, or a new one), and delete all three marker lines. VS
   Code also offers buttons above the conflict; try them if you like.
5. Finish the merge:

```bash
git add songs.txt
git commit --no-edit
git log --oneline --graph
```

**Why question:** git merged plenty of changes on its own today. Why can
it not just pick one of the two titles here? What would go wrong if it
guessed?

*Your answer here.*

## Stretch 7: a pull request on your own repo

Back in your **real** course repo. This is how almost all work on GitHub
is reviewed and merged, and it is the branch workflow from today with a
web page in the middle.

1. Make a branch called `add-cheatsheet`.
2. On it, write `GIT-CHEATSHEET.md` in the root of your repo: the
   commands from these stretches, each with one line **in your own
   words** about what it is for.
3. Commit it, then push the **branch**, not `main`:

```bash
git push -u origin add-cheatsheet
```

4. Open your repo on github.com. There will be a yellow banner offering
   **Compare & pull request**. Click it, write a title and a sentence
   about what the change is, and create the pull request.
5. Click the **Files changed** tab and read your own change as a
   reviewer would. Leave yourself a comment on one line.
6. Click **Merge pull request** (keep the default, "Create a merge
   commit"), confirm, and then **Delete branch**.
7. GitHub now has a commit your laptop does not. Bring your copy up to
   date and tidy up:

```bash
git switch main
git pull
git branch -d add-cheatsheet
git log --oneline --graph -5
```

Is `GIT-CHEATSHEET.md` on `main` now? Can you see the merge in the graph?
