Notebook 01 cleaned the file one cell at a time. That is the right way to
*find* the steps, and the wrong way to *keep* them: next week you would
have to run thirty cells in the right order to get clean data back.

So turn it into two functions, in the shape session 10 builds on:

```python
raw = load_messy("data/messy/plays_messy.csv")
plays = clean_plays(raw)
```

1. `load_messy(path)` reads the file and returns it **untouched**. Use
   `keep_default_na=False`, so pandas guesses nothing and every decision
   about what is missing is yours.
2. `clean_plays(df)` takes that and returns a cleaned copy with **exactly
   the columns, column order and types of `data/clean/plays.csv`**. Then
   everything you wrote in session 8 works on it unchanged.
3. Check it against the clean file. The checks are written for you; when
   they all pass, you are done.

The session 3 rule still holds: the functions **return**, and only the
cells at the bottom print.

The decisions from notebook 01, as a checklist. Because `load_messy` keeps

what each blank means.
