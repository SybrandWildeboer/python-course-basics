## Setup

Finds the `project/` folder whichever folder this notebook was opened from,
and imports `pipeline.py` from it.

```python
import importlib
import sys
from pathlib import Path

import pandas as pd

def find_project(start):
    """Return the folder holding pipeline.py, looking in start and above it."""
    for folder in [start, *start.parents]:
        if (folder / "pipeline.py").exists():
            return folder
        if (folder / "project" / "pipeline.py").exists():
            return folder / "project"
    raise FileNotFoundError(f"no pipeline.py found in or above {start}")


PROJECT = find_project(Path.cwd())

sys.path.insert(0, str(PROJECT))
import pipeline

importlib.reload(pipeline)      # pick up changes you made to pipeline.py since
print("project folder:", PROJECT)
print("input file:    ", pipeline.INPUT)
```
